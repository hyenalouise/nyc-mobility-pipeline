```python
taxi_zones_df = (
    spark.read
    .option("header", True)
    .csv(
        "/Volumes/ftw-week-08/00-source/group_a_source/taxi_zones/taxi_zone_lookup.csv"
    )
)

display(taxi_zones_df)
```

```python
taxi_zones_df.createOrReplaceTempView("taxi_zones")
```

```python
%sql
SELECT
    LocationID,
    COUNT(*) AS duplicate_count
FROM taxi_zones
GROUP BY LocationID
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    SUM(CASE WHEN LocationID IS NULL THEN 1 ELSE 0 END) AS null_locationid,
    SUM(CASE WHEN Borough IS NULL THEN 1 ELSE 0 END) AS null_borough,
    SUM(CASE WHEN Zone IS NULL THEN 1 ELSE 0 END) AS null_zone,
    SUM(CASE WHEN service_zone IS NULL THEN 1 ELSE 0 END) AS null_service_zone
FROM taxi_zones;
```

```python
%sql
SELECT DISTINCT Borough
FROM taxi_zones
ORDER BY Borough;
```

```python
%sql
SELECT *
FROM taxi_zones
WHERE Borough = 'Unknown'
   OR Zone LIKE '%Unknown%'
   OR Zone LIKE '%Outside%';
```
